```python
from langgraph.graph import StateGraph, START, END

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_groq import ChatGroq
from langchain_core.messages import SystemMessage, HumanMessage

from typing import TypedDict, Literal, Annotated
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
generator_model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash"
)

review_model = ChatGroq(
    model="llama-3.3-70b-versatile"
)

optimizer_model = ChatGroq(
    model="llama-3.1-8b-instant"
)
```

```python
from pydantic import BaseModel, Field

class TweetEvaluation(BaseModel) :
    evaluation: Literal["approved", "needs_improvement"] = Field(..., description="Final evaluation result.")
    feedback: str = Field(..., description="feedback for the tweet.")
```

```python
structured_evaluator_llm = review_model.with_structured_output(TweetEvaluation)
```

```python
class TweetState(TypedDict) :
    topic : str
    tweet : str
    evaluation : Literal["approved", "needs_improvement"]
    feedback : str
    iteration : int
    max_iteration : int
```

```python
def generate_tweet(state : TweetState) :
    print("\n========== GENERATE ==========")
    #prompt
    messages = [
        SystemMessage(content="You are a funny and clever Twitter/X influencer."),
        HumanMessage(content=f"""
Write a short, original, and hilarious tweet on topic: "{state['topic']}".

Rules: 
- Do not use question-answer format.
- Max 120 characters.
- Use obeservational humor, irony, sarcasm, or cultural references.
- Think in meme logic, punchlines, or relatable takes.
- Use simple, day to day english 
""")
    ]

    #send generator_llm
    response = generator_model.invoke(messages).content

    #return response
    return {'tweet' : response}
```

```python
def evaluate_tweet(state: TweetState):
    print("\n========== EVALUATE ==========")
    print("Iteration:", state["iteration"])

    messages = [
        SystemMessage(
            content="""
You are a ruthless, no-laugh-given Twitter/X content critic.

Your only responsibility is to evaluate tweets.

Judge every tweet like an experienced viral content reviewer.

Your evaluation must be based on the following criteria:

1. Originality
- Is the joke fresh?
- Does it avoid overused internet jokes?
- Would an average Twitter user feel they've seen it before?

2. Humor
- Is it genuinely funny?
- Does it create surprise, irony, sarcasm, absurdity, or relatability?
- Does it actually make someone smile?

3. Punchiness
- Is every word useful?
- Is it concise?
- Does it end with a strong punchline?

4. Virality
- Would people like, repost, or quote this tweet?
- Is it likely to become a meme or spark discussion?

5. Format
- Must look like a real Twitter/X post.
- NOT question-answer format.
- NOT a setup-punchline joke.
- Maximum 120 characters.
- Natural everyday English.

Automatically reject the tweet if:

- It is written as a Q&A.
- It feels like a traditional joke.
- It is generic.
- It is too long.
- It explains the joke.
- It ends with weak filler.
- It repeats the prompt.
- It sounds AI generated.

Respond ONLY in this exact format.

evaluation: approved

feedback:
Explain why the tweet works.


OR


evaluation: needs_improvement

feedback:
Explain exactly what should be improved.
Do NOT rewrite the tweet.
"""
        ),

        HumanMessage(
            content=f"""
Evaluate the following tweet.

Tweet:

{state['tweet']}
"""
        )
    ]

    response = structured_evaluator_llm.invoke(messages)

    return {'evaluation' : response.evaluation, 'feedback' : response.feedback}
```

```python
def optimize_tweet(state : TweetState) : 
    print("\n========== OPTIMIZE ==========")
    print("Iteration:", state["iteration"])

    messages = [
        SystemMessage(
            content="""
    You are an elite Twitter/X copywriter and content optimizer.

    Your only responsibility is to improve tweets using reviewer feedback.

    Never generate a completely new idea unless absolutely necessary.
    Keep the original intent, topic, and joke intact while making it stronger.

    Your goals are:

    1. Improve Humor
    - Increase surprise, irony, sarcasm, wit, or relatability.
    - Make the punchline stronger.
    - Remove predictable jokes.

    2. Improve Originality
    - Replace clichés with fresh observations.
    - Avoid overused internet humor.
    - Make the tweet feel unique.

    3. Improve Punchiness
    - Remove unnecessary words.
    - Make every word earn its place.
    - End with a memorable punchline.

    4. Improve Virality
    - Make people want to like, repost, or quote it.
    - Add meme-like energy when appropriate.
    - Keep it relatable and shareable.

    5. Improve Format
    - Maximum 120 characters.
    - Natural everyday English.
    - NOT question-answer format.
    - NOT setup-punchline joke format.
    - Do NOT explain the joke.
    - Do NOT use hashtags unless they genuinely improve the tweet.
    - Do NOT use emojis unless they significantly improve humor.

    Rules:
    - Preserve the original topic.
    - Preserve the original joke whenever possible.
    - Follow the reviewer's feedback carefully.
    - Return ONLY the improved tweet.
    - Do not include explanations, notes, or markdown.
    """
        ),

        HumanMessage(
            content=f"""
    Original Tweet:

    {state["tweet"]}

    Reviewer Feedback:

    {state["feedback"]}

    Rewrite the tweet by addressing every issue mentioned in the feedback.
    """
        )
    ]

    response = optimizer_model.invoke(messages).content
    iteration = state["iteration"] + 1

    return {'tweet' : response, 'iteration' : iteration}
```

```python
def route_condition(state: TweetState):

    print("\n========== ROUTER ==========")
    print(state)

    if state["iteration"] < 3:
        print("GO -> optimize")
        return "need_improvement"

    if state["evaluation"] == "approved":
        print("GO -> END")
        return "approved"

    if state["iteration"] >= state["max_iteration"]:
        print("GO -> END (MAX)")
        return "approved"

    print("GO -> optimize")

    return "need_improvement"
```

```python
graph = StateGraph(TweetState)

graph.add_node('generate', generate_tweet)
graph.add_node('evaluate', evaluate_tweet)
graph.add_node('optimize', optimize_tweet)

graph.add_edge(START, 'generate')
graph.add_edge('generate', 'evaluate')
graph.add_conditional_edges('evaluate', route_condition, {'approved' : END, 'need_improvement' : 'optimize'})
graph.add_edge('optimize', 'evaluate')

workflow = graph.compile()
```

```python
from langgraph.graph import StateGraph, START, END
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash"
)
```

```python
# Create State
class LLMState(TypedDict) :
    question : str
    answer : str
```

```python
def llm_qa(state : LLMState) -> LLMState : 
    # extract the question from state
    question = state['question']

    # Form a prompt 
    prompt = f'Answer the following question { question }'

    # Ask that question to the LLM [For getting only content add .content at last after (prompt).content]
    answer = model.invoke(prompt).content

    # Update the anwer in the state 
    state['answer'] = answer

    return state
```

```python
graph = StateGraph(LLMState)

graph.add_node('llm_que', llm_qa)

graph.add_edge(START, 'llm_que')
graph.add_edge('llm_que', END)

workflow = graph.compile()
```

```python
initial_state = {'question' : 'How far is moon from the earth ?'}

final_state = workflow.invoke(initial_state)

print(final_state['answer'])
```
